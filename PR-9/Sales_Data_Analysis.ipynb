```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


class SalesDataAnalyzer:
    def __init__(self, file_path=None):
        self.data = None
        self.last_figure = None
        if file_path:
            self.load_data(file_path)

    def __del__(self):
        print("SalesDataAnalyzer object destroyed. Cleanup complete.")

    def load_data(self, file_path):
        try:
            self.data = pd.read_csv(file_path)
            print("Dataset loaded successfully!")
        except Exception as e:
            print("Error loading dataset:", e)

    def explore_data(self):
        if self.data is None:
            print("No data loaded. Please load a dataset first.")
            return
        while True:
            print("\n== Explore Data ==")
            print("1. Display the first 5 rows")
            print("2. Display the last 5 rows")
            print("3. Display column names")
            print("4. Display data types")
            print("5. Display basic info")
            print("6. Back to main menu")
            choice = input("Enter your choice: ")
            if choice == "1":
                print(self.data.head())
            elif choice == "2":
                print(self.data.tail())
            elif choice == "3":
                print(self.data.columns.tolist())
            elif choice == "4":
                print(self.data.dtypes)
            elif choice == "5":
                print(self.data.info())
            elif choice == "6":
                break
            else:
                print("Invalid choice. Try again.")

    def handle_missing_data(self):
        if self.data is None:
            print("No data loaded. Please load a dataset first.")
            return
        while True:
            print("\n== Handle Missing Data ==")
            print("1. Display rows with missing values")
            print("2. Fill missing values with mean")
            print("3. Drop rows with missing values")
            print("4. Replace missing values with a specific value")
            print("5. Back to main menu")
            choice = input("Enter your choice: ")
            if choice == "1":
                missing_rows = self.data[self.data.isnull().any(axis=1)]
                if missing_rows.empty:
                    print("No missing values found in the dataset!")
                else:
                    print(missing_rows)
            elif choice == "2":
                numeric_cols = self.data.select_dtypes(include=np.number).columns
                for col in numeric_cols:
                    self.data[col] = self.data[col].fillna(self.data[col].mean())
                print("Missing values filled with mean successfully!")
            elif choice == "3":
                self.data = self.data.dropna()
                print("Rows with missing values dropped successfully!")
            elif choice == "4":
                value = input("Enter the value to replace missing values with: ")
                self.data = self.data.fillna(value)
                print("Missing values replaced successfully!")
            elif choice == "5":
                break
            else:
                print("Invalid choice. Try again.")

    def numpy_operations(self):
        column = input("Enter numeric column name to convert to Numpy array: ")
        if column not in self.data.columns:
            print("Column not found.")
            return
        arr = self.data[column].to_numpy()
        print("Numpy array:")
        print(arr)
        print("First 5 elements:", arr[:5])
        print("Last 3 elements:", arr[-3:])
        print("Element at index 0:", arr[0])
        try:
            start = int(input("Enter start index for slicing: "))
            end = int(input("Enter end index for slicing: "))
            print("Sliced array:", arr[start:end])
        except Exception as e:
            print("Invalid input:", e)

    def mathematical_operations(self):
        column = input("Enter numeric column name for mathematical operations: ")
        if column not in self.data.columns:
            print("Column not found.")
            return
        arr = self.data[column].to_numpy()
        print("Sum:", np.sum(arr))
        print("Mean:", np.mean(arr))
        print("Max:", np.max(arr))
        print("Min:", np.min(arr))
        try:
            num = float(input("Enter a number to add to each element: "))
            print("After addition:", arr + num)
            num2 = float(input("Enter a number to multiply each element by: "))
            print("After multiplication:", arr * num2)
        except Exception as e:
            print("Invalid input:", e)

    def combine_data(self):
        file_path = input("Enter file path of another CSV to combine: ")
        try:
            other_df = pd.read_csv(file_path)
            method = input("Enter combine method (concat/merge): ")
            if method == "concat":
                self.data = pd.concat([self.data, other_df], ignore_index=True)
                print("Data combined using concat successfully!")
            elif method == "merge":
                key = input("Enter common column name to merge on: ")
                self.data = pd.merge(self.data, other_df, on=key)
                print("Data combined using merge successfully!")
            else:
                print("Invalid method selected.")
        except Exception as e:
            print("Error combining data:", e)

    def split_data(self):
        column = input("Enter column name to split data by: ")
        if column not in self.data.columns:
            print("Column not found.")
            return
        groups = self.data.groupby(column)
        for name, group in groups:
            print("\nGroup:", name)
            print(group)

    def search_sort_filter(self):
        while True:
            print("\n== Search, Sort and Filter ==")
            print("1. Search")
            print("2. Sort")
            print("3. Filter")
            print("4. Back")
            choice = input("Enter your choice: ")
            if choice == "1":
                column = input("Enter column name to search in: ")
                value = input("Enter value to search: ")
                if column not in self.data.columns:
                    print("Column not found.")
                    continue
                result = self.data[self.data[column].astype(str) == value]
                print(result)
            elif choice == "2":
                column = input("Enter column name to sort by: ")
                order = input("Ascending or Descending (A/D): ")
                if column not in self.data.columns:
                    print("Column not found.")
                    continue
                ascending = True if order.upper() == "A" else False
                sorted_data = self.data.sort_values(by=column, ascending=ascending)
                print(sorted_data)
            elif choice == "3":
                column = input("Enter column name to filter by: ")
                value = input("Enter value to filter for: ")
                if column not in self.data.columns:
                    print("Column not found.")
                    continue
                filtered_data = self.data[self.data[column].astype(str) == value]
                print(filtered_data)
            elif choice == "4":
                break
            else:
                print("Invalid choice. Try again.")

    def aggregate_functions(self):
        column = input("Enter numeric column name for aggregation: ")
        if column not in self.data.columns:
            print("Column not found.")
            return
        print("Sum:", self.data[column].sum())
        print("Mean:", self.data[column].mean())
        print("Count:", self.data[column].count())
        print("Max:", self.data[column].max())
        print("Min:", self.data[column].min())
        group_col = input("Enter column name to group by (or leave blank to skip): ")
        if group_col:
            if group_col not in self.data.columns:
                print("Column not found.")
                return
            print(self.data.groupby(group_col)[column].sum())

    def create_pivot_table(self):
        index_col = input("Enter column name for pivot table index: ")
        columns_col = input("Enter column name for pivot table columns: ")
        values_col = input("Enter column name for pivot table values: ")
        try:
            pivot = pd.pivot_table(self.data, index=index_col, columns=columns_col, values=values_col, aggfunc="sum")
            print(pivot)
        except Exception as e:
            print("Error creating pivot table:", e)

    def dataframe_operations(self):
        if self.data is None:
            print("No data loaded. Please load a dataset first.")
            return
        while True:
            print("\n== Perform DataFrame Operations ==")
            print("1. Numpy Array Creation, Indexing and Slicing")
            print("2. Mathematical Operations")
            print("3. Combine Data")
            print("4. Split Data")
            print("5. Search, Sort and Filter")
            print("6. Aggregate Functions")
            print("7. Create Pivot Table")
            print("8. Back to main menu")
            choice = input("Enter your choice: ")
            if choice == "1":
                self.numpy_operations()
            elif choice == "2":
                self.mathematical_operations()
            elif choice == "3":
                self.combine_data()
            elif choice == "4":
                self.split_data()
            elif choice == "5":
                self.search_sort_filter()
            elif choice == "6":
                self.aggregate_functions()
            elif choice == "7":
                self.create_pivot_table()
            elif choice == "8":
                break
            else:
                print("Invalid choice. Try again.")

    def statistical_analysis(self):
        if self.data is None:
            print("No data loaded. Please load a dataset first.")
            return
        column = input("Enter numeric column name for statistical analysis: ")
        if column not in self.data.columns:
            print("Column not found.")
            return
        print(self.data[column].describe())
        print("Standard Deviation:", self.data[column].std())
        print("Variance:", self.data[column].var())
        print("25th Percentile:", self.data[column].quantile(0.25))
        print("50th Percentile:", self.data[column].quantile(0.5))
        print("75th Percentile:", self.data[column].quantile(0.75))

    def visualize_data(self):
        if self.data is None:
            print("No data loaded. Please load a dataset first.")
            return
        while True:
            print("\n== Data Visualization ==")
            print("1. Bar Plot")
            print("2. Line Plot")
            print("3. Scatter Plot")
            print("4. Pie Chart")
            print("5. Histogram")
            print("6. Stack Plot")
            print("7. Heatmap")
            print("8. Box Plot")
            print("9. Back to main menu")
            choice = input("Enter your choice: ")
            if choice == "1":
                x_col = input("Enter x-axis column name: ")
                y_col = input("Enter y-axis column name: ")
                fig, ax = plt.subplots()
                ax.bar(self.data[x_col], self.data[y_col])
                ax.set_xlabel(x_col)
                ax.set_ylabel(y_col)
                ax.set_title("Bar Plot")
                plt.xticks(rotation=45)
                plt.tight_layout()
                self.last_figure = fig
                plt.show()
                print("Bar plot displayed successfully!")
            elif choice == "2":
                x_col = input("Enter x-axis column name: ")
                y_col = input("Enter y-axis column name: ")
                fig, ax = plt.subplots()
                ax.plot(self.data[x_col], self.data[y_col])
                ax.set_xlabel(x_col)
                ax.set_ylabel(y_col)
                ax.set_title("Line Plot")
                plt.xticks(rotation=45)
                plt.tight_layout()
                self.last_figure = fig
                plt.show()
                print("Line plot displayed successfully!")
            elif choice == "3":
                x_col = input("Enter x-axis column name: ")
                y_col = input("Enter y-axis column name: ")
                fig, ax = plt.subplots()
                ax.scatter(self.data[x_col], self.data[y_col])
                ax.set_xlabel(x_col)
                ax.set_ylabel(y_col)
                ax.set_title("Scatter Plot")
                self.last_figure = fig
                plt.show()
                print("Scatter plot displayed successfully!")
            elif choice == "4":
                column = input("Enter categorical column name for pie chart: ")
                fig, ax = plt.subplots()
                counts = self.data[column].value_counts()
                ax.pie(counts, labels=counts.index, autopct="%1.1f%%")
                ax.set_title("Pie Chart")
                self.last_figure = fig
                plt.show()
                print("Pie chart displayed successfully!")
            elif choice == "5":
                column = input("Enter numeric column name for histogram: ")
                fig, ax = plt.subplots()
                ax.hist(self.data[column], bins=10)
                ax.set_xlabel(column)
                ax.set_ylabel("Frequency")
                ax.set_title("Histogram")
                self.last_figure = fig
                plt.show()
                print("Histogram displayed successfully!")
            elif choice == "6":
                x_col = input("Enter x-axis column name: ")
                y_cols = input("Enter y-axis column names separated by comma: ")
                y_list = [c.strip() for c in y_cols.split(",")]
                fig, ax = plt.subplots()
                ax.stackplot(self.data[x_col], *[self.data[c] for c in y_list], labels=y_list)
                ax.legend(loc="upper left")
                ax.set_title("Stack Plot")
                self.last_figure = fig
                plt.show()
                print("Stack plot displayed successfully!")
            elif choice == "7":
                fig, ax = plt.subplots()
                numeric_data = self.data.select_dtypes(include=np.number)
                sns.heatmap(numeric_data.corr(), annot=True, ax=ax)
                ax.set_title("Heatmap")
                self.last_figure = fig
                plt.show()
                print("Heatmap displayed successfully!")
            elif choice == "8":
                column = input("Enter numeric column name for box plot: ")
                fig, ax = plt.subplots()
                sns.boxplot(y=self.data[column], ax=ax)
                ax.set_title("Box Plot")
                self.last_figure = fig
                plt.show()
                print("Box plot displayed successfully!")
            elif choice == "9":
                break
            else:
                print("Invalid choice. Try again.")

    def save_visualization(self):
        if self.last_figure is None:
            print("No visualization available to save. Please generate a plot first.")
            return
        filename = input("Enter file name to save the plot (e.g., scatter_plot.png): ")
        try:
            self.last_figure.savefig(filename)
            print("Visualization saved as", filename, "successfully!")
        except Exception as e:
            print("Error saving visualization:", e)


def main():
    analyzer = SalesDataAnalyzer()
    while True:
        print("\n========== Data Analysis & Visualization Program ==========")
        print("Please select an option:")
        print("1. Load Dataset")
        print("2. Explore Data")
        print("3. Perform DataFrame Operations")
        print("4. Handle Missing Data")
        print("5. Generate Descriptive Statistics")
        print("6. Data Visualization")
        print("7. Save Visualization")
        print("8. Exit")
        print("=============================================================")
        choice = input("\nEnter your choice: ")
        if choice == "1":
            file_path = input("Enter the path of the dataset (CSV file): ")
            analyzer.load_data(file_path)
        elif choice == "2":
            analyzer.explore_data()
        elif choice == "3":
            analyzer.dataframe_operations()
        elif choice == "4":
            analyzer.handle_missing_data()
        elif choice == "5":
            analyzer.statistical_analysis()
        elif choice == "6":
            analyzer.visualize_data()
        elif choice == "7":
            analyzer.save_visualization()
        elif choice == "8":
            print("Exiting the program. Goodbye!")
            break
        else:
            print("Invalid choice. Please try again.")


main()
```
